```python
import pandas as pd

# 1. Load fund master and NAV history using relative path from notebooks folder
fund_master = pd.read_csv("../Data/Raw/01_fund_master.csv")
nav_history = pd.read_csv("../Data/Raw/02_nav_history.csv")

# Identify scheme code column name (handles both 'scheme_code' and 'amfi_code')
master_code_col = 'scheme_code' if 'scheme_code' in fund_master.columns else 'amfi_code'
nav_code_col = 'scheme_code' if 'scheme_code' in nav_history.columns else 'amfi_code'

# 2. Extract unique scheme codes from both datasets
master_codes = set(fund_master[master_code_col].dropna().unique())
nav_codes = set(nav_history[nav_code_col].dropna().unique())

# 3. Validate using Pandas merge/indicator to identify unmapped codes
merged = pd.merge(
    fund_master[[master_code_col]], 
    nav_history[[nav_code_col]].drop_duplicates(), 
    left_on=master_code_col,
    right_on=nav_code_col,
    how='left', 
    indicator=True
)

missing_in_nav = merged[merged['_merge'] == 'left_only']

# 4. Generate Data Quality Summary Report (Assignment Deliverable)
print("==========================================")
print("         DATA QUALITY REPORT             ")
print("==========================================")
print(f"Total Unique Codes in Fund Master : {len(master_codes)}")
print(f"Total Unique Codes in NAV History : {len(nav_codes)}")
print("------------------------------------------")

if missing_in_nav.empty:
    print("STATUS: PASSED ✔")
    print("SUMMARY: Every scheme code in fund_master exists in nav_history.")
else:
    print("STATUS: ANOMALY DETECTED ✘")
    print(f"SUMMARY: Found {len(missing_in_nav)} scheme code(s) in fund_master missing from nav_history:")
    print(missing_in_nav[[master_code_col]])
print("==========================================")
```
